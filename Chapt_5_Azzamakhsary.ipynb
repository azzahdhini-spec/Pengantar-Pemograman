```python
import pandas as pd
from scipy.stats.mstats import gmean, hmean
import matplotlib.pyplot as plt

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

a_mean = my_dataset.Lu.mean()
g_mean = gmean(my_dataset['Lu'])
h_mean = hmean(my_dataset['Lu'])

print ('-------')
print ('arithmetic mean')
print ("{0:.1f} [ppm]".format(a_mean))
print ('-------')

print ('geometric mean')
print ("{0:.1f} [ppm]".format(g_mean))
print ('-------')

print ('harmonic mean')
print ("{0:.1f} [ppm]".format(h_mean))
print ('-------')

fig, ax = plt.subplots()
ax.hist(my_dataset.Lu, bins='auto', density=True, edgecolor='k', label='Measurements Hist', alpha=0.8) 
ax.axvline(a_mean, color='blue', label='Arithmetic mean', linewidth=4)
ax.axvline(g_mean, color='yellow',  label='Geometric mean', linewidth=4)
ax.axvline(h_mean, color='pink',  label='Harmonic mean', linewidth=4)
ax.set_xlabel('Lu [ppm]')
ax.set_ylabel('Probability density')
ax.legend()

'''
Output:
-------
arithmetic mean
365.4 [ppm]
-------
geometric mean
348.6 [ppm]
-------
harmonic mean
333.8 [ppm]
-------
'''
```

```python
import pandas as pd
import matplotlib.pyplot as plt

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

variance =  my_dataset['Rb'].var()
stddev =  my_dataset['Rb'].std()

print ('-------')
print ('Variance')
print("{0:.0f} [square ppm]".format(variance))
print ('-------')
print ('Standard Deviation')
print("{0:.0f} [ppm]".format(stddev))
print ('-------')

fig, ax = plt.subplots()
ax.hist(my_dataset.Rb, bins= 20, density = True, edgecolor='k', label='Measurements Hist') 
ax.axvline(my_dataset['Rb'].mean()-stddev, color='red', label=r'mean - 1$\sigma$', linewidth=4)
ax.axvline(my_dataset['Rb'].mean()+stddev, color='yellow', label=r'mean + 1$\sigma$', linewidth=4)
ax.axvspan(my_dataset['Rb'].mean()-stddev, my_dataset['Rb'].mean()+stddev, alpha=0.1, color='pink', label=r'mean $\pm$ 1$\sigma$')
ax.set_xlabel('Rb [ppm]')
ax.set_ylabel('Probability density')
ax.legend()

'''
Output:
-------
Variance
14021 [square ppm]
-------
Standard Deviation
118 [ppm]
-------
'''
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

q1 = np.percentile(my_dataset.Y, 25, method = 'midpoint') 
q3 = np.percentile(my_dataset.Y, 75, method = 'midpoint') 
  
iqr = q3 - q1 # Interquaritle range (IQR)  

print ('-------')
print ('Interquaritle range (IQR): {0:.0f} [ppm]'.format(iqr))
print ('-------')

fig, ax = plt.subplots()
ax.hist(my_dataset.Y, bins='auto', density=True, edgecolor='k', label='Measurements Hist') 
ax.axvline(q1, color='purple', label='Q1', linewidth=4)
ax.axvline(q3, color='yellow', label='Q3', linewidth=4)
ax.axvspan(q1, q3, alpha=0.1, color='red', label='Interquaritle range (IQR)')
ax.set_xlabel('Y [ppm]')
ax.set_ylabel('Probability density')
ax.legend()

'''
Output:
-------
Interquaritle range (IQR): 164 [ppm]
-------
'''
```

```python
import numpy as np
from scipy.stats import skew

a_mean = my_dataset.Cs.mean()
median = my_dataset.Cs.median()
hist, bin_edges = np.histogram(my_dataset['Cs'], bins=20, density=True)
modal_value = (bin_edges[hist.argmax()] + bin_edges[hist.argmax()+1])/2
standard_deviation = my_dataset['Cs'].std()

a1 = (a_mean - modal_value) / standard_deviation 
a2 = 3 * (a_mean - median) / standard_deviation
g1 = skew(my_dataset['Cs'])

print ('-------')
print ("Pearson's first coefficient of skewness: {:.2f}".format(a1))
print ("Pearson's 2nd moment of skewness: {:.2f}".format(a2))
print ("Fisher-Pearson's coefficient of skewness: {:.2f}".format(g1))
print ('-------')

'''
Output:
-------
Pearson's first coefficient of skewness: 0.74
Pearson's 2nd moment of skewness: 0.66
Fisher-Pearson's coefficient of skewness: 1.26
-------
'''
```
